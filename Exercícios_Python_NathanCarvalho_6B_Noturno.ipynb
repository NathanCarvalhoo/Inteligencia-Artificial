**Exercício 12**

Conte o número de ocorrências de cada caractere em uma string fornecida.

```python
#variável "palavra" solicita que o usuário digite uma palavra.
palavra = str(input("Digite uma palavra: "))
#Adicionamos um dicionário vazio que será usado para armazenar a quantidade de cada caractere.
caractere = {}
#Iniciamos um loop para identificar cada letra da variável "palavra".
for i in palavra:
  #Durante o loop a condição verifica se a letra já foi armazenada ao dicionário.
  if i in caractere:
    #caractere é o dicionário onde armazena as contagens das letras;
    #O "i" é substituido por cada letra da palavra digitada pelo usuário;
    #O operador "+= 1" soma +1 a cada letra repetida da variável "palavra" e armazena o resultado de volta no dicionário.
    caractere[i] += 1
  #Utilizamos o else para caso a letra ainda não esteja armazenada ao dicionário.
  else:
    #O "caractere[i] = 1" é responsável por adicionar a letra que ainda não foi adicionada ao dicionário.
    caractere[i] = 1
#Durante o 2° loop o "i" será substituido por cada letra da variável "palavra";
#O "Count" identifica a quantidade que cada letra foi contada e armazena novamente ao dicionário.
for i, count in caractere.items():
    #Printamos a letra e a quantidade de cada letra da palavra.
    print(f"{i} : {count}")
```

**Exercício 17**

Conte a frequência de cada palavra em uma string fornecida pelo
usuário.

```python
#A variável "frase" solicita que o usuário digite uma frase.
frase = str(input("Digite uma frase: "))
#Utilizamos o ".split", responsável por separar cada palavra da frase.
palavras = frase.split()
#Criamos um dicionário chamado "frequencia";
#O dicionário é responsável por armazenar as palavras da frase como uma chave e quantidade de vezes que essa palavra aparece na frase.
frequencia = {}
#Iniciamos um loop para identificar cada palavra da frase.
for i in palavras:
  #Dentro do loop adicionamos uma condição, responsável por identificar se a palavra já foi armazenada no dicionário "frequencia".
  if i in frequencia:
    #O operador "+= 1" soma +1 a cada palavra repetida da frase e armazena o resultado de volta no dicionário.
    frequencia[i] += 1
    #Caso a palavra não seja adicionada.
  else:
    #Ela então acaba sendo adicionada, e sua quantidade passa a ser 1.
    frequencia[i] = 1
#O segundo loop é responsável por identificar a palavra e a frequência de cada uma delas armazenadas no dicionário "frequencia";
#O "i" seria cada uma das palavras da frase;
#O "count" seria a quantidade (frequência) que ele aparece na frase.
for i, count in frequencia.items():
  #Por fim, imprimimos a palavra e a frequência de cada uma delas.
  print(f"{i} : {count}")
```

**Exercício 21**

Implemente a busca binária para encontrar um elemento em uma
lista ordenada.

```python
#Definimos uma função recebendo "lista" e "alvo".
def busca_binaria(lista, alvo):
  #A variável "inicio" recebe valor de 0, sendo literalmente o início.
  inicio = 0
  #A variável "fim" é a última posição da lista.
  fim = len(lista) - 1
  #O loop continua rodando até que a posição de início ultrapasse o fim.
  while inicio <= fim:
    #O "meio" soma a posição inicial e final, e divide por 2 para descobrir a posição exata do meio.
    meio = (inicio + fim) // 2
    #Se o elemento que está no meio for o "alvo".
    if lista[meio] == alvo:
      #Retorna "meio".
      return meio
    #Se o número do meio for menor que o "alvo".
    elif lista[meio] < alvo:
      #O "inicio" passa a ser o número "meio + 1".
      inicio = meio + 1
    #Se o número do meio for maior que o "alvo".
    else:
      #O "fim" passa a ser o número "meio - 1".
      fim = meio - 1
  #Caso nenhuma das anteriores resolva, retorna "-1".
  return -1
#Criamos uma lista ordenada.
minha_lista = [3, 7, 14, 19, 23, 28, 35, 47, 51, 70]
#A variável "alvo" solicita que o usuário digite o elemento que deseja buscar.
alvo = int(input("Digite o elemento que deseja buscar: "))
#O "resultado" será o resultado da busca_binaria, podendo ser a posição do número ou -1 .
resultado = busca_binaria(minha_lista, alvo)
#Caso resultado seja diferente de -1.
if resultado != -1:
    #Printamos que o elemento foi encontrado e onde ele está.
    print(f"Elemento encontrado no índice: {resultado}")
#se não
else:
    #Imprime que o elemento não está na lista
    print("Elemento não está na lista.")
```

```python
#Definimos um limite de caracteres para o qtd_cpf.
qtd_cpf = 14
#Criamos uma variável true, informando que ela é válida.
numero = True
#A variável "palavra" solicita que o usuário digite uma palavra.
cpf = str(input("Digite o CPF: "))
#A primeira condição identifica se a quantidade de caracteres é diferente da determinada pelo "qtd_cpf".
if (len(cpf) != qtd_cpf):
  #Caso seja diferente, o cpf será "inválido".
  numero = False
#Caso a quantidade seja certa, a condição identifica se os "." e "-" estão nas posições corretas (3,7,11).
elif(cpf[3] != '.' or cpf[7] != '.' or cpf[11] != '-'):
  #Caso esteja incorreto, o cpf será inválido.
  numero = False
#Se todas condições forem corretas.
else:
  #Inicia-se um loop do cpf.
  for i in range(qtd_cpf):
    #Dentro do loop a condição verifica se onde deveria ser "." e "-" há números.
    if((i != 3)) and (i != 7) and (i !=11):
      #A variável "c" identifica cada numero, pontos e hífen digitado pelo usuário.
      c = cpf[i]
    #Esta condição verifica se variável "c" não é um digito.
    if (not c.isdigit()):
      #Caso não seja, o "numero" será inválido.
      numero = False
#Se depois de todas as condições o "numero" for válido.
if (numero):
  #Printa que o cpf é válido.
  print(f"O CPF é válido")
#Caso contrário.
else:
  #Printa que é inválido.
  print(f"O CPF é inválido")
```

Desenvolva um jogo simples da forca onde o usuário deve adivinhar
uma palavra.

```python
#Importamos o módulo random.
import random
#Definimos a função jogar_forca.
def jogar_forca():
    #Criamos uma variável "palavras" sendo uma delas a escolhida para ser a resposta certa.
    palavras = ['PYTHON', 'COMPUTADOR', 'ALGORITMO', 'NOTEBOOK', 'TECLADO', 'SOFTWARE']
    #"palavrasecreta" seleciona uma das palavras da lista para ser a resposta.
    palavrasecreta = random.choice(palavras)
    #"letrascertas" armazena todas as letras certas digitadas pelo usuário.
    letrascertas = set()
    #"letraserradas" armazena todas as letras erradas digitadas pelo usuário.
    letraserradas = set()
    #"tentativas" são as quantidades de chances que o usuário tem para acertar.
    tentativas = 5
    #O jogo roda enquanto o usuário estiver com "tentativas" maior que 0.
    while tentativas > 0:
        #"palavraoculta" literalmente oculta as letras não adivinhadas e substitui por "_".
        palavraoculta = [letra if letra in letrascertas else "_" for letra in palavrasecreta]
        #Imprime a visualização do jogo;
        #".join(palavraoculta)" pega a lista de "_" e letras ocultas e separa cada uma delas.
        print("Palavra:", " ".join(palavraoculta))
        #Imprime a quantidade de tentativas que o usuário ainda tem.
        print(f"Tentativas restantes: {tentativas}")
        #No loop há condições, caso a letra dada pelo usuário não esteja na palavra do jogo.
        if letraserradas:
          #Ele imprime a letra errada e também as que já foram tentadas.
          print("Letras já tentadas e erradas:", " - ".join(letraserradas))
        #Se não houver mais "_" na lista.
        if "_" not in palavraoculta:
          #O jogador vencer e imprime uma mensagem de parabéns.
          print(f"Parabéns! Você adivinhou a palavra: {palavrasecreta}")
          #Finaliza o loop.
          break
        #A variável "chute" solicita que o usuário digite uma letra;
        #".strip" remove espaços em branco;
        #".upper" transformam as letras em maiúsculas.
        chute = input("Digite uma letra: ").strip().upper()
        #adicionamos uma condição onde verificamos se foi digitada mais de uma letra ou nenhuma;
        #"chute.isalpha()" verifica se o que foi digitado é uma letra.
        if len(chute) != 1 or not chute.isalpha():
          #Caso condição seja verdadeira, ele pede para que digite apenas uma letra do alfabeto.
          print("Por favor, digite apenas uma letra do alfabeto.")
          #O "continue" retorna o loop, para o usuário tentar novamente.
          continue
        #Essa condição verifica se a letra já foi digitada.
        if chute in letrascertas or chute in letraserradas:
          #Caso condição seja verdadeira, ele pede para que digite uma outra letra do alfabeto .
          print("Você já tentou essa letra, escolha outra.")
          #O "continue" retorna o loop, para o usuário tentar novamente.
          continue
        #Essa condição verifica se a letra digitada está na palavra secreta.
        if chute in palavrasecreta:
          #Se a condição for verdadeira, printamos que a letra está na palavra.
          print("Essa letra está na palavra.")
          #Adicionamos a letra dentro do (set) de "letrascertas".
          letrascertas.add(chute)
        #se não.
        else:
          #printamos que a letra não está na palavra
          print("Essa letra não está na palavra.")
          #Adicionamos a letra dentro do (set) de "letraserradas".
          letraserradas.add(chute)
          #A tentativa diminui toda vez que usuário erra a letra.
          tentativas -= 1
    #Se a quantidade de tentativas for igual a 0.
    if tentativas == 0:
      #Printamos que o usuário perdeu o jogo.
      print("Fim de jogo!")
      #E printamos a resposta correta.
      print(f"A palavra secreta era: {palavrasecreta}")
#Condiçao que executa todo o jogo.
if __name__ == "__main__":
    jogar_forca()
```

**Exercício 31**

Implemente o algoritmo de ordenação por bolha (Bubble Sort).

```python
#Definimos uma função para bubblesort que recebe uma "lista".
def bubblesort(lista):
  #A variável "n" conta a quantidade de números existentes na lista.
  n = len(lista)
  #Iniciamos um loop de 0 até n-1.
  for i in range(n):
    #Criamos variável de controle, identificando se ela já está ordenada
    trocou = False
    #Nesta condição o loop vai diminuindo.
    for j in range(0, n - i - 1):
      #condição onde verificamos se o número "j" é maior que o número direita dele.
      if lista[j] > lista[j + 1]:
        #Caso seja verdade, ele inverte as posições dos dois números.
        lista[j], lista[j + 1] = lista[j + 1], lista[j]
        #Se a troca foi feita, mudamos a variável controle para "True".
        trocou = True
    #Se nenhuma troca ocorreu nesta passagem, a lista já está ordenada.
    if not trocou:
      #Encerrariamos o loop.
      break
  #Retorna a lista organizada.
  return lista
#Executa o bubble sort
if __name__ == "__main__":
  #Criamos a variável "numeros" armazenando uma lista desordenada.
  numeros = [64, 39, 82, 57, 22, 10, 3]
  #Imprimimos a lista desordenada.
  print(f"Lista desordenada: {numeros}")
  #Recebemos a lista "numeros" e ordenamos baseado na função bubblesort.
  numerosordenados = bubblesort(numeros)
  #Printamos a lista ordenada.
  print(f"Lista ordenada: {numerosordenados}")
```

**Exercício 32**

Crie um jogo da velha para dois jogadores.

```python
#Definimos uma função para imprimirtabuleiro que recebe uma lista "tabuleiro".
def imprimirtabuleiro(tabuleiro):
  #Printamos 5 linhas para ajudar na visualização.
  print(f" {tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]} ")
  print("-----------")
  print(f" {tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]} ")
  print("------------")
  print(f" {tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]} ")
#criamos uma segunda função chamado "verificarvitoria".
def verificarvitoria(tabuleiro, jogador):
    combinacoes = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]
    #inicia um loop.
    for combo in combinacoes:
      #A conndição identifica se os três espaços possuem o mesmo símbolo.
      if tabuleiro[combo[0]] == tabuleiro[combo[1]] == tabuleiro[combo[2]] == jogador:
        #Se sim, retorna true.
        return True
    #Caso contrário, retorna false.
    return False
#Definimos uma função jogodavelha.
def jogodavelha():
    #adicionamos uma lista com nove espaços.
    tabuleiro = [str(i) for i in range(1, 10)]
    #Definimos o jogador como "X".
    jogador = 'X'
    #Controla a rodada.
    jogadas = 0
    #Criamos loop para rodar o jogo
    while True:
        imprimirtabuleiro(tabuleiro)
        #A variável "escolha" solicita que o usuário digite uma posição;
        escolha = input(f"Vez do jogador {jogador}, escolha uma posição 1 a 9: ").strip()
        #Condição que verifica se o usuário digitou corretamente.
        if not escolha.isdigit() or int(escolha) < 1 or int(escolha) > 9:
            print("Erro. Digite um número de 1 a 9.")
            #O "continue" retorna o loop, para o usuário tentar novamente.
            continue
        #Define a escolha do usuário
        indice = int(escolha) - 1
        #Checa se a posição já está ocupada por 'X' ou 'O'
        if tabuleiro[indice] in ['X', 'O']:
            print("Essa posição já está ocupada, escolha outra.")
            #O "continue" retorna o loop, para o usuário tentar novamente.
            continue
        #Identifica a posição e substitui pelo "X" ou "O"
        tabuleiro[indice] = jogador
        jogadas += 1
        #Condição onde verifica a vitória
        if verificarvitoria(tabuleiro, jogador):
            imprimirtabuleiro(tabuleiro)
            print(f"Parabéns! O jogador {jogador} venceu!")
            #Encerrariamos o loop.
            break
        #Condição onde verifica empate
        #Se todos os 9 posições forem completadas
        if jogadas == 9:
            imprimirtabuleiro(tabuleiro)
            print("Velha! O jogo terminou empatado.")
            #Encerrariamos o loop.
            break
        #Altera a vez do jogador .
        jogador = "O" if jogador == "X" else "X"
#Condição onde executa o jogo
if __name__ == "__main__":
    jogodavelha()
```

Gere todos os números primos até um número n fornecido.

```python
#variável "numero" solicita que o usuário digite um número inteiro.
numero = input("Digite um número inteiro positivo: ")
#Condição que verifica se a entrada é um número positivo.
if numero.isdigit() and int(numero) > 0:
    n = int(numero)
    #Adiconamos uma lista para armazenar os números primos encontrados
    numerosprimos = []
    #O loop inicia em 2 até n para verificar cada número
    for num in range(2, n + 1):
        # Assume que o número é primo inicialmente
        primo = True
        #O loop verifica divisibilidade de 2 até a metade do número
        #Se for divisível por algum número nesse intervalo, não é primo
        for i in range(2, int(num**0.5) + 1):
            if num % i == 0:
                primo = False
                #interrompe o loop
                break
        #Se o número permaneceu primo após as verificações, adiciona à lista
        if primo:
            numerosprimos.append(num)
    #Imprime os números primos encontrados
    print(f"Os números primos até {n} são: {numerosprimos}")
else:
    #Mensagem de erro para entrada inválida
    print("Por favor, digite um número inteiro positivo válido.")
```

**Exercício 43**

Converta unidades de comprimento (por exemplo, metros para
centímetros).

```python
#Adicionamos uma tabela de conversão.
fatoresparametro = {"mm": 0.001, "cm": 0.01, "m": 1.0, "km": 1000.0  }
#Printamos ao usuário que pode converter.
print("Unidades: mm, cm, m, km")
#variável "palavra" solicita que o usuário digite uma palavra.
unidadeorigem = input("Digite a unidade de origem: ").strip().lower()
unidadeesperada = input("Digite a unidade de esperada: ").strip().lower()
#Verifica se as unidades digitadas existem no nosso dicionário.
if unidadeorigem in fatoresparametro and unidadeesperada in fatoresparametro:
    #Captura o valor numérico e troca vírgula por ponto.
    entradavalor = input(f"Digite o valor em '{unidadeorigem}' para converter: ").strip().replace(',', '.')
    try:
        #Tenta converter o texto digitado para um número decimal.
        valor = float(entradavalor)
        #Converte o valor original para Metros.
        valoremmetros = valor * fatoresparametro[unidadeorigem]
        #Converte os Metros para a unidade esperada.
        resultado = valoremmetros / fatoresparametro[unidadeesperada]
        # O :.4f formata o número para exibir até 4 casas decimais.
        print(f"Resultado: {valor} {unidadeorigem} = {resultado:.6f} {unidadeesperada}")
    #Se o usuário digitar letras em vez de número.
    except ValueError:
        print("Digite um valor numérico")
#Se não.
else:
    print("Unidade não reconhecida.")
```

**Exercício 44**

Gere a sequência de Fibonacci até um determinado número de
termos.

```python
#Definimos uma função para gerarfibonacci.
def gerarfibonacci(n):
    #Condição para entradas menores ou iguais a 1.
    if n <= 0:
        return []
    elif n == 1:
        return [0]
    #Inicializa a lista com os dois primeiros números da sequência.
    sequencia = [0, 1]
    #O loop começa a partir do 3° termo até n.
    for _ in range(2, n):
        #Soma os dois últimos números da lista.
        proximonumero = sequencia[-1] + sequencia[-2]
        sequencia.append(proximonumero)
    return sequencia
entrada = input("Quantos termos você deseja gerar? ")
#A condição valida para garantir que o usuário digitou um número.
if entrada.isdigit():
    quantidade = int(entrada)
    #Chama a função passando a quantidade desejada.
    resultado = gerarfibonacci(quantidade)
    #Printa a quantidade de termos
    print(f"Os primeiros {quantidade} termos são:")
    #Separa os números por vírgula.
    print(", ".join(str(numero) for numero in resultado))
#Se não.
else:
  #printamos erro
  print("Digite apenas números positivos.")
```

**Exercício 45**

Classifique os números de uma lista como pares ou ímpares.

```python
#Definimos uma função "classificarparesimpares"
def classificarparesimpares(numeros):
    #Separa uma lista de números em duas listas: pares e ímpares.
    pares = []
    impares = []
    #Adicionamos uma condição
    for numero in numeros:
      #Se restando da divisão numero e 2 for 0
      if numero % 2 == 0:
        #Adiciona a lista par
        pares.append(numero)
      #se não.
      else:
        #Adiciona na lista ímpar
        impares.append(numero)
    #retorna listas pares e impares.
    return pares, impares
#Adicionamos uma lista com dados inseridos nela.
lista = [12, 7, 3, 15, 8, 20, 1]
#Como a função retorna duas listas, podemos separá-los em duas variáveis
listapares, listaimpares = classificarparesimpares(lista)
#Printamos a lista anterios e as listas separadas por par e ímpar.
print(f"Lista: {lista}")
print(f"Lista pares: {listapares}")
print(f"Lista ímpares: {listaimpares}")
```

**Exercício 46**

Calcule a média e o desvio padrão de uma lista de números.

```python
#Importamos a biblioteca.
import statistics
#Adicionamos uma variável "dados" com uma lista de números.
dados = [9, 15, 23, 11, 37, 18, 22]
#Calculamos a média da lista.
media = statistics.mean(dados)
#Calculamos o desvio padrão da lista.
desviopadrao = statistics.stdev(dados)
#printamos a lista com os dados, a média e o desvio padrão.
print(f"Lista: {dados}")
print(f"Média: {media:.2f}")
print(f"Desvio Padrão: {desviopadrao:.2f}")
```

**Exercício 47**

Conte quantas vezes cada número aparece em uma lista.

```python
#Criamos uma variável "entrada" e solicitamos que o usuário digite os números.
entrada = input("Digite números separados por espaço (ex: 1 2 3 4 5): ")
try:
    #Transformamos o texto digitado em uma lista de números.
    numeros = [int(x) for x in entrada.split()]
    #Criamos um dicionário vazio para guardar as contagens.
    contagem = {}
    #Iniciamos um loop.
    for numero in numeros:
        #Dentro do loop, adicionamos uma condição;
        #Se o número já está no dicionário, soma +1 na contagem dele.
        if numero in contagem:
            contagem[numero] += 1
        #Se não, se for a primeira vez que o número aparece, a contagem dele começa em 1.
        else:
            contagem[numero] = 1
    #Loop para exibir o resultado em formato de tabela.
    for numero, qtd in contagem.items():
        print(f"{numero:} | {qtd}")
#Caso usuário não digite números inteiros. Printamos erro.
except ValueError:
    print("Erro, digite apenas números inteiros.")
```

**Exercício 48**

Implemente o algoritmo de ordenação por inserção (Insertion Sort).

```python
#Definimos uma função para "insercao".
def insercao(lista):
    #O loop se inicia a partir do 1, pois o 0 já constitui uma "lista ordenada".
    for i in range(1, len(lista)):
        #A "chave" é o valor que queremos inserir.
        chave = lista[i]
        #A variável "j" é o valor anterior à chave
        j = i - 1
        #Durante o loop alteramos os valores que são maiores que a "chave".
        while j >= 0 and lista[j] > chave:
          #Caso a lista[j] for maior que a "chave", mudamos a posição para a direita.
          lista[j + 1] = lista[j]
          #E o "j" se compara com a "chave".
          j -= 1
        #Após o loop acabar, significa que achamos a posição correta logo.
        lista[j + 1] = chave
    #retornamos a lista ordenada.
    return lista
#Condição onde executa todo código.
if __name__ == "__main__":
    #Criamos uma lista de números aleatórias.
    numeros = [39, 63, 17, 55, 73]
    #Exibe a lista aleatória
    print(f"Lista: {numeros}")
    #Usamos a função "numeros" para organizar a lista.
    numerosordenados = insercao(numeros)
    #Printamos a lista ordenada.
    print(f"Lista ordenada: {numerosordenados}")
```

**Exercício 50**

Calcule a frequência de cada letra em uma string fornecida pelo
usuário.

```python
#Importamos a classe Counter do módulo collections.
from collections import Counter
#variável "texto" solicita que o usuário digite um texto;
#Formatamos com ".lower()" para que o texto fique em minúsculo.
texto = input("Digite um texto: ").lower()
#"char for char in texto" identifica cada caractere da frase;
#".isalpha()" identifica apenas o que é letra.
frequenciacounter = Counter(char for char in texto if char.isalpha())
#O "sorted(sorted(frequenciacounter.keys()" organiza as letras em ordem alfabética.
for letra in sorted(frequenciacounter.keys()):
    #Printamos a letra em maiúsculo e a quantidade de vezes que apareceu.
    print(f"{letra.upper()} | {frequenciacounter[letra]}")
#O método "most_common(3)" devolve as 3 letras mais digitadas no texto.
for letra, qtd in frequenciacounter.most_common(3):
  #Printamos as letras que mais se repetem no texto.
  print(f"{letra.upper()} = {qtd} vezes")
```
