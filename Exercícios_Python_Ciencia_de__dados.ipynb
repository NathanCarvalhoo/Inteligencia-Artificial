**Exercício 19**

Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado).

```python
import numpy as np
array1 = np.random.randint(1, 10, size=20)
array2 = np.random.randint(1, 10, size=20)
produtoescalar = np.dot(array1, array2)
normaarray1 = np.linalg.norm(array1)
nomaarray2 = np.linalg.norm(array2)
cosseno = produtoescalar / (normaarray1 * nomaarray2)
angulo = np.degrees(np.arccos(cosseno))
print(f"Array 1: {array1}")
print(f"Array 2: {array2}")
print(f"O ângulo entre eles é de: {angulo:.3f}")
```

**Exercício 1**

Calcule a integral da função $f(x) = x^2 \sin(x)$ no intervalo de 0 a $\pi$.

```python
import numpy as np
from scipy.integrate import quad
def f(x):
    return (x**2) * np.sin(x)
resultado, f = quad(f, 0, np.pi)
print(f"Resultado da integral: {resultado:.4f}")
```

**Exercício 2**

Resolva a equação diferencial $\frac{dy}{dx} = x - y$, com $y(0) = 1$, no intervalo de 0 a 10.

```python
from scipy import integrate
def f(x, y):
    return x - y
sol = integrate.solve_ivp(f, [0, 10], [1], t_eval=[0, 1, 5, 10])
print(sol.y[0])
```

**Exercício 3**

Encontre as raízes da função $f(x) = x^4 - 3x^3 + 2$ utilizando SciPy.

```python
import numpy as np
from scipy.optimize import fsolve
def f(x):
    return x**4 - 3 * x**3 + 2
raizes = fsolve(f, [-1.0, 1.0, 3.0])
print(np.unique(np.round(raizes, 4)))
```

**Exercício 4**

Use a função de otimização de SciPy para maximizar $f(x) = -x^2 + 4x + 1$.

```python
from scipy.optimize import minimize_scalar
def menosf(x):
    return -(-x**2 + 4 * x + 1)
resultado = minimize_scalar(menosf)
print(f"Ponto máximo:{resultado.x:.2f}")
print(f"Valor máximo:f(x)={-resultado.fun:.2f}")
```

**Exercício 5**

Calcule a transformada de Fourier de um sinal composto de $\sin(2\pi t)$ e $\cos(4\pi t)$.

```python
import numpy as np
from scipy.fft import fft
t = np.linspace(0, 1, 100, endpoint=False)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)
transformada = fft(sinal)
print(np.round(transformada[:4], 2))
```

**Exercício 9**

Ajuste uma curva quadrática aos pontos (1, 2), (2, 4), (3, 5), (4, 7), (5, 8).

```python
import numpy as np
from scipy import optimize
x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 7, 8])
def curvaquadratica(x, a, b, c):
    return a * x**2 + b * x + c
coeficientes, f = optimize.curve_fit(curvaquadratica, x, y)
print(f"Equação ajustada: y = {coeficientes[0]:.4f}x^2 + {coeficientes[1]:.4f}x + {coeficientes[2]:.4f}")
```

**Exercício 10**

Calcule a integral dupla da função $f(x, y) = xy + x^2$ no intervalo [0, 2] para $x$ e [0, 1] para $y$.

**Exercício 6**

Plote dois gráficos de linha em subplots diferentes: $y_1 = e^x$ e $y_2 = \log(x)$.

**Exercício 8**

Plote um gráfico de linha com intervalos de confiança (sombreamento) para a função $y = \sin(x)$.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y = np.sin(x)
plt.plot(x, y)
plt.fill_between(x, y - 0.2, y + 0.2, alpha=0.3)
plt.show()
```

**Exercício 11**

Crie um gráfico de barras empilhadas para três categorias com três grupos de valores.

```python
import numpy as np
import matplotlib.pyplot as plt
categorias = ["A", "B", "C"]
g1 = np.array([5, 10, 15])
g2 = np.array([8, 12, 6])
g3 = np.array([4, 7, 9])
plt.bar(categorias, g1)
plt.bar(categorias, g2, bottom=g1)
plt.bar(categorias, g3, bottom=g1 + g2)
plt.show()
```

**Exercício 18**

Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor.

```python
import matplotlib.pyplot as plt
categorias = ["A", "B", "C", "D"]
valores = [15, -10, 25, -5]
cores = ["green" if v >= 0 else "red" for v in valores]
plt.bar(categorias, valores, color=cores)
plt.show()
```

**Exercício 19**

Plote um gráfico de linha com duas séries e adicione uma legenda para diferenciá-las.

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.plot(x, y1, label="Seno")
plt.plot(x, y2, label="Cosseno")
plt.legend()
plt.show()
```

**Exercício 20**

Adicione uma grade com espaçamento personalizado a um gráfico de linha para $y = x^2$

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
x = np.linspace(-5, 5, 100)
y = x**2
fig, ax = plt.subplots()
ax.plot(x, y)
ax.xaxis.set_major_locator(ticker.MultipleLocator(1))
ax.yaxis.set_major_locator(ticker.MultipleLocator(5))
ax.grid(True)
plt.show()
```
