1. Criação de uma lista vazia

2. Primeiro Loop (for): Treinando modelos padrão

**.fit(X_train, y_train)**: O algoritmo analisa a matriz de medidas (**X_train**) e cruza com o gabarito (**y_train**) para extrair os padrões matemáticos de cada espécie.

**.predict(X_test)**: Passamos as medidas dos 40% de dados ocultos (**X_test**). O modelo gera um vetor com suas previsões (**y_pred**), sem  ter visto o resultado daquelas flores.

**accuracy_score**: Calcula o percentual total de acertos

**recall_score**: Como temos 3 classes (espécies) diferentes, o parâmetro average='macro' instrui a função a calcular a capacidade de acerto para a classe 0, depois para a 1, depois para a 2, e tirar a média exata disso.
